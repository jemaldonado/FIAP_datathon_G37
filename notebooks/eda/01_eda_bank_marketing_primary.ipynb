# EDA — Bank Marketing (Primary Dataset)

**Source**: henriqueyamahata/bank-marketing  
**Role**: Primary — base for processed layer and bandit simulation  
**Target**: `y` — term deposit subscription (binary)  

Run ETL first: `python -m datathon.etl.bank_marketing_primary`

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style='whitegrid')
DATA = Path('../../data')
```

## 1. Shape and Schema

```python
df = pd.read_parquet(DATA / 'processed/bank_marketing_primary.parquet')
print(f'Shape: {df.shape}')
df.dtypes
```

```python
df.head()
```

## 2. Missing Values

```python
missing = df.isnull().sum()
missing[missing > 0].sort_values(ascending=False)
```

## 3. Target Distribution

```python
fig, ax = plt.subplots()
df['y'].value_counts().plot(kind='bar', ax=ax)
ax.set_title('Target distribution (y=1: subscribed)')
ax.set_xlabel('y')
ax.set_ylabel('count')
plt.tight_layout()
plt.show()
print(df['y'].value_counts(normalize=True))
```

## 4. Numeric Feature Distributions

```python
num_cols = df.select_dtypes(include='number').columns.tolist()
num_cols = [c for c in num_cols if c != 'y']
df[num_cols].hist(bins=30, figsize=(16, 10))
plt.tight_layout()
plt.show()
```

## 5. Categorical Feature Distributions

```python
cat_cols = df.select_dtypes(include='object').columns.tolist()
for col in cat_cols:
    fig, ax = plt.subplots(figsize=(8, 3))
    df[col].value_counts().plot(kind='bar', ax=ax)
    ax.set_title(col)
    plt.tight_layout()
    plt.show()
```

## 6. Correlation Heatmap (Numeric Features)

```python
fig, ax = plt.subplots(figsize=(12, 10))
sns.heatmap(df[num_cols + ['y']].corr(), annot=True, fmt='.2f', ax=ax)
plt.tight_layout()
plt.show()
```

## 7. Conversion Rate by Categorical Feature

```python
for col in cat_cols:
    rates = df.groupby(col)['y'].mean().sort_values(ascending=False)
    fig, ax = plt.subplots(figsize=(8, 3))
    rates.plot(kind='bar', ax=ax)
    ax.set_title(f'Conversion rate by {col}')
    ax.set_ylabel('P(y=1)')
    plt.tight_layout()
    plt.show()
```

## 8. Leakage Audit

```python
leakage_cols = ['duration']
present = [c for c in leakage_cols if c in df.columns]
if present:
    print(f'WARNING: leakage columns present: {present}')
else:
    print('OK: no leakage columns found in processed dataset')
```
